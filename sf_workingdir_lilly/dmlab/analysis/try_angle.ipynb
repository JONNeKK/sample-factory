```python
import numpy as np 
```

```python
X=np.random.randn(10,50)
```

```python
X=X.T
```

```python
X.shape
```

```python
subspace=X[:,[0]]
ortho_components=[]
for i in range(1,X.shape[1]):
    current_x=X[:,i]
    current_x=current_x/np.linalg.norm(current_x)
    projection=subspace @ np.linalg.pinv(subspace) @ current_x
    orthogonal_component=current_x-projection
    ortho_components.append(orthogonal_component)
    ortho_normal_component=orthogonal_component/np.linalg.norm(orthogonal_component)
    subspace=np.column_stack((subspace,ortho_normal_component))
```

```python
import matplotlib.pyplot as plt
tmp=np.array(ortho_components)
tmp=np.linalg.norm(tmp,axis=1)
plt.plot(tmp)
```
