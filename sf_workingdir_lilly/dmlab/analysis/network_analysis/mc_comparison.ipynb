```python
weight_trials = [1, 3, 4, 5, 6, 7, 8, 9, 11, 12, 13, 14, 15, 17, 18, 22, 23, 24, 27, 28, 29, 30, 33, 35, 36, 38, 39, 40, 41, 42, 43, 45, 47, 49]

seeds = [1111,2222,3333, 4444,5555]
```

```python
print(mc_measures['orth_vec_thresh_crossed_not_norm'])
print(mc_measures_lora['orth_vec_thresh_crossed_not_norm'])
```

```python
orth_thresh_lora = np.zeros((len(seeds), len(weight_trials)))
singular_components_krylov_lora = np.zeros((len(seeds), len(weight_trials)))
print(orth_thresh_lora.shape)

for t_i, t in enumerate(weight_trials):
    for s_i,seed in enumerate(seeds):
        orth_thresh_lora[s_i,t_i] = mc_measures_lora['orth_vec_thresh_crossed_not_norm'][t][seed]
        singular_components_krylov_lora[s_i,t_i] = mc_measures_lora['MC_singular_components_svd'][t][seed]
```

```python
singular_components_krylov_lora_mean_seeds = singular_components_krylov_lora.mean(axis=0)
orth_thresh_lora_mean_seeds = orth_thresh_lora.mean(axis=0)
```

```python
print(len(singular_components_krylov_lora_mean_seeds))
singular_components_krylov = np.asarray(mc_measures['MC_singular_components_svd'])
orth_thresh= np.asarray(mc_measures['orth_vec_thresh_crossed_not_norm'])
print(len(singular_components_krylov))
```

#### test whether LORA increses MC

```python
# H0: E(d)=0 there is no difference between the two measures
# H1: E(d)>0 LoRA increases MC

d = singular_components_krylov_lora_mean_seeds - singular_components_krylov
n = d.size

print(f"n = {n}")
print(f"Mean difference (LoRA - base): {d.mean():.6g}")
print(f"Median difference: {np.median(d):.6g}")

# paired t-test
t_stat, p_two = stats.ttest_rel(singular_components_krylov_lora_mean_seeds, singular_components_krylov, nan_policy="omit")
p_one = p_two / 2 if t_stat > 0 else 1 - p_two / 2
print("\nPaired t-test (H1: LoRA > base)")
print(f"t = {t_stat:.4f}, p(one-sided) = {p_one:.4g}, p(two-sided) = {p_two:.4g}")
```

```python
try:
    w_stat, p_w = stats.wilcoxon(d, alternative="greater")  # one-sided
    print("\nWilcoxon signed-rank (H1: LoRA > base)")
    print(f"W = {w_stat:.4f}, p(one-sided) = {p_w:.4g}")
except ValueError as e:
    print("\nWilcoxon failed (often due to all differences being zero):", e)
```

```python
d = singular_components_krylov_lora_mean_seeds - singular_components_krylov
#d_trial = np.nanmean(d, axis=0) 
t_stat, p_two = stats.ttest_1samp(d, 0.0, nan_policy="omit")
print(t_stat, p_two)
print("mean diff (LoRA-base):", np.nanmean(d))
```

```python
d = orth_thresh_lora_mean_seeds - orth_thresh
#d_trial = np.nanmean(d, axis=0) 
t_stat, p_two = stats.ttest_1samp(d, 0.0, nan_policy="omit")
print(t_stat, p_two)
print("mean diff (LoRA-base):", np.nanmean(d))
```
