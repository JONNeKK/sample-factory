```python
from sf_workingdir_lilly.dmlab.custom_esn_analysis import esn_pca, get_hidden_states, visualize_pca_results
import matplotlib.pyplot as plt
import numpy as np
```

### PCA of the 1st weights

```python

spectral_radius = [0.3, 0.4, 0.5, 0.6, 0.7, 0.8, 0.9, 0.92, 0.95, 0.97, 0.99]

for sr in spectral_radius:
    hs = get_hidden_states(70, sr, 16, 8, 64)  
    td = esn_pca(hs)
    print(td)
    print(td.shape)

    #visualize_pca_results(td)
    plt.figure()
    plt.plot(td[:,0],td[:,1])

    step = 1 
    x, y = td[::step, 0], td[::step, 1]
    u, v = np.diff(x), np.diff(y)
    plt.quiver(x[:-1], y[:-1], u, v,
            angles='xy', scale_units='xy', scale=1,
            width=0.003, alpha=0.8)


    plt.xlabel('PC1')
    plt.ylabel('PC2')
    plt.legend()
    plt.title(f'spectral radius {sr}')
    plt.show()
```
